```python
from matplotlib import pyplot as plt
```

```python
plt.figure(figsize=(8, 4), dpi=300)
x = range(2013, 2026)
y = [1, 1, 1, 2, 6, 12, 18, 22, 21, 31, 31, 38, 42]  # 2025 total updated July 31, 2025
bars = plt.bar(x, y, color="tab:blue")
plt.xticks(x)
ax = plt.gca()
ax.bar_label(bars, color="tab:blue")
ax.yaxis.grid(True, color="#DDDDDD")
ax.set_axisbelow(True)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.ylabel("Publications on Microhaplotypes")
plt.savefig("fig1a.png", bbox_inches="tight")
plt.savefig("fig1a.jpg", bbox_inches="tight")
plt.savefig("fig1a.pdf", bbox_inches="tight")
```

```python
plt.figure(figsize=(8,5), dpi=300)
x = list(range(2015, 2026))
kidd2014 = [9, 15, 26, 42, 63, 83, 96, 116, 136, 151, 153]
kidd2017 = [0, 0, 5, 18, 37, 54, 65, 81, 93, 105, 106]
oldoni2019 = [0, 0, 0, 0, 17, 36, 52, 73, 90, 109, 112]
plt.plot(x, kidd2014, color="tab:blue", marker="o", label="Kidd $\it{et\ al.}$2014$^{[x]}$")
plt.plot(x[2:], kidd2017[2:], color="tab:orange", marker="s", label="Kidd $\it{et\ al.}$2017$^{[y]}$")
plt.plot(x[4:], oldoni2019[4:], color="tab:green", marker="*", label="Oldoni & Podini 2019$^{[z]}$", markersize=8)
plt.xticks(x)
ax = plt.gca()
ax.yaxis.grid(True, color="#DDDDDD")
ax.set_axisbelow(True)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.ylabel("Cumulative Publications")
plt.legend(loc="upper left")
for year, num in zip(x, kidd2014):
    plt.text(year, num + 2.5, str(num), ha='center', va='bottom', fontsize=8, color="tab:blue")
for year, num in zip(x, kidd2017):
    if num == 0:
        continue
    if year < 2024:
        plt.text(year, num + 2.5, str(num), ha='center', va='bottom', fontsize=8, color="tab:orange")
    else:
        plt.text(year, num - 3.25, str(num), ha='center', va='top', fontsize=8, color="tab:orange")
for year, num in zip(x, oldoni2019):
    if num == 0:
        continue
    if year < 2024:
        plt.text(year, num - 3.25, str(num), ha='center', va='top', fontsize=8, color="tab:green")
    else:
        plt.text(year, num + 2.5, str(num), ha='center', va='bottom', fontsize=8, color="tab:green")
plt.savefig("fig1b.png", bbox_inches="tight")
plt.savefig("fig1b.jpg", bbox_inches="tight")
plt.savefig("fig1b.pdf", bbox_inches="tight")
```
